```python
"""Exercise 3: Book tickets while seats remain (5 points · about 12 min)
There are 5 tickets available. Use while to ask how many tickets the next person wants.

0 stops taking bookings.
A request larger than the number remaining prints Not enough tickets. Keep the remaining number unchanged and ask again.
Otherwise, subtract the request and print Remaining: followed by the new number.
When no tickets remain, stop automatically. Print Booking closed once after the loop.
Assume each entry is a whole number, zero or above.

Checks: requests 2, 4, 3 leave 3 tickets, reject 4, then leave 0 and stop without another question. Request 0 first: only Booking closed. Request 5 first: Remaining: 0, then Booking closed."""
```

```python
tickets = 5
while tickets > 0:
    request= int(input("How many tickets do you want?"))
    if request == 0:
        break
    if request  > tickets:
        print("Not enough tickets")
    else:
        tickets= tickets -request
        print("Remaining:", tickets)
print("Booking closed")

```

```python
Exercise 5: Pair stock records with zip (4 points · about 8 min)
Each item and its stock quantity are at the same position in the two lists. Use zip() and a condition to print Refill: followed by the item name when its quantity is below 2.

Checks: the supplied data prints Refill: Coffee and Refill: Milk. A quantity of exactly 2 does not need a refill. Change stock to [2, 3, 2]: no output.

Extra: also print how many units are needed to reach 2.
```

```python
#Exercise 5:
items = ["Tea", "Coffee", "Milk"]
stock = [2, 1, 0]
for item, quantity in zip(items, stock):
    if quantity < 2:
        print ("Refill:", item)

# or print ("Refill:", item, quantity)
```

```python
## Optional challenge: handle a negative request
Extend Exercise 3. A negative ticket request should print `Enter 0 or more` and ask again, without changing the remaining tickets. You can use `continue`.

**Check:** `-2`, `2`, `0` should reject the negative request, leave 3 tickets after the booking, then close. Optional work is not needed for full marks.
```

```python
tickets = 5
while tickets > 0:
    request= int(input("How many tickets do you want?"))
    if request < 0:
        print("Enter 0 or more")
        continue
    if request == 0:
        break
    if request  > tickets:
        print("Not enough tickets")
    else:
        tickets= tickets -request
        print("Remaining:", tickets)

print("Booking closed")
```
