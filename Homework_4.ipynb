```python
#Exercise 5
stock = {"Uno": 2, "Azul": 1, "Dixit": 0}
requests = ["Uno", "Dixit", "Twister", "Uno", "Uno", "Azul"]

for game in requests:
    if game not in stock:
        print("Unknown game",game)
    elif stock[game] == 0:
        print("Unavaiable",game)
    else:
        stock[game] -= 1
        print("Borrowed", game)
print(stock)
```

```python
stock = {"Uno": 2, "Azul": 1, "Dixit": 0}
requests = ["Uno", "Dixit", "Twister", "Uno", "Uno", "Azul"]
borrowed = []
for game in requests:
    if game not in stock:
        print("Unknown game",game)
    elif stock[game] == 0:
        print("Unavaiable",game)
    else:
        stock[game] -= 1
        borrowed.append(game)
print(borrowed)

print(len(set(borrowed)))


```
